```python
import os, math, numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim

from torchvision import datasets, transforms
from torch.utils.data import DataLoader, random_split, WeightedRandomSampler
from sklearn.metrics import (
    f1_score, roc_auc_score, average_precision_score,
    classification_report, confusion_matrix,
    roc_curve, precision_recall_curve
)

from torch.amp import autocast, GradScaler
from tqdm import tqdm
from PIL import Image
import matplotlib.pyplot as plt
import seaborn as sns

# ================= CONFIG =================
DATA_PATH = "/kaggle/input/datasets/paultimothymooney/chest-xray-pneumonia/chest_xray"
EPOCHS = 50
BATCH_SIZE = 64
LR = 1e-4
VAL_SPLIT = 0.15
SEED = 42

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ================= TRANSFORMS =================
train_tf = transforms.Compose([
    transforms.Grayscale(1),
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.RandomAffine(0, translate=(0.08, 0.08)),
    transforms.ColorJitter(0.2, 0.2),
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5])
])

val_tf = transforms.Compose([
    transforms.Grayscale(1),
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5])
])

# ================= DATA =================
full_train = datasets.ImageFolder(os.path.join(DATA_PATH, "train"))

val_size = int(len(full_train) * VAL_SPLIT)
train_size = len(full_train) - val_size

train_raw, val_raw = random_split(
    full_train, [train_size, val_size],
    generator=torch.Generator().manual_seed(SEED)
)

class Wrap(torch.utils.data.Dataset):
    def __init__(self, subset, tf):
        self.subset = subset
        self.tf = tf
    def __len__(self):
        return len(self.subset)
    def __getitem__(self, i):
        path, label = self.subset.dataset.samples[self.subset.indices[i]]
        img = Image.open(path).convert('L')
        return self.tf(img), label

train_ds = Wrap(train_raw, train_tf)
val_ds   = Wrap(val_raw,   val_tf)

# ✅ TEST SET (NO LEAKAGE)
test_ds = datasets.ImageFolder(os.path.join(DATA_PATH, "test"), val_tf)

# ================= SAMPLER =================
targets = [full_train.samples[i][1] for i in train_raw.indices]
class_counts = np.bincount(targets)
class_weights = 1.0 / class_counts
sample_weights = [class_weights[t] for t in targets]

sampler = WeightedRandomSampler(sample_weights, len(sample_weights), replacement=True)

# ================= LOADERS =================
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=sampler, num_workers=4)
val_loader   = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=4)
test_loader  = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=4)

# ================= LOSS =================
class FocalLoss(nn.Module):
    def __init__(self, alpha=0.4, gamma=2.0):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma
    def forward(self, inputs, targets):
        ce_loss = F.cross_entropy(inputs, targets, reduction='none')
        pt = torch.exp(-ce_loss)
        return (self.alpha * (1 - pt) ** self.gamma * ce_loss).mean()

criterion = FocalLoss()

# ================= MODEL =================
class CNNStem(nn.Module):
    def __init__(self, out_dim=256):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(64, 128, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(128, out_dim, 3, stride=2, padding=1), nn.ReLU()
        )
    def forward(self, x):
        x = self.net(x)
        B, C, H, W = x.shape
        return x.permute(0,2,3,1).reshape(B, H*W, C)

class SimpleBlock(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.fc1 = nn.Linear(dim, dim*4)
        self.fc2 = nn.Linear(dim*4, dim)
        self.norm = nn.LayerNorm(dim)
    def forward(self, x):
        return self.norm(x + self.fc2(F.gelu(self.fc1(x))))

class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.stem = CNNStem(256)
        self.blocks = nn.Sequential(*[SimpleBlock(256) for _ in range(4)])
        self.head = nn.Sequential(
            nn.Linear(512,256),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(256,2)
        )
    def forward(self,x):
        x = self.stem(x)
        x = self.blocks(x)
        mean = x.mean(1)
        maxv = x.max(1)[0]
        x = torch.cat([mean,maxv], dim=1)
        return self.head(x)

model = Model().to(device)

optimizer = optim.AdamW(model.parameters(), lr=LR)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
scaler = GradScaler()

# ================= TRACKING =================
val_losses, val_f1s, roc_aucs, pr_aucs = [], [], [], []

# ================= TRAIN =================
best_f1 = 0

for epoch in range(EPOCHS):
    model.train()
    for x,y in tqdm(train_loader):
        x,y = x.to(device), y.to(device)
        optimizer.zero_grad()

        with autocast('cuda'):
            out = model(x)
            loss = criterion(out,y)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

    scheduler.step()

    # ===== VALIDATION =====
    model.eval()
    probs, labels = [], []
    val_loss = 0

    with torch.no_grad():
        for x,y in val_loader:
            x,y = x.to(device), y.to(device)
            out = model(x)
            loss = criterion(out,y)
            val_loss += loss.item()

            p = torch.softmax(out,1)[:,1].cpu().numpy()
            probs.extend(p)
            labels.extend(y.cpu().numpy())

    probs = np.array(probs)
    labels = np.array(labels)
    preds = (probs > 0.5).astype(int)

    f1 = f1_score(labels, preds)
    roc_auc = roc_auc_score(labels, probs)
    pr_auc = average_precision_score(labels, probs)

    val_losses.append(val_loss / len(val_loader))
    val_f1s.append(f1)
    roc_aucs.append(roc_auc)
    pr_aucs.append(pr_auc)

    print(f"Epoch {epoch+1} | F1:{f1:.4f} ROC:{roc_auc:.4f}")

    if f1 > best_f1:
        best_f1 = f1
        torch.save(model.state_dict(), "best_model.pth")

# ================= THRESHOLD =================
model.load_state_dict(torch.load("best_model.pth"))
model.eval()

val_probs, val_labels = [], []

with torch.no_grad():
    for x,y in val_loader:
        x = x.to(device)
        out = model(x)
        p = torch.softmax(out,1)[:,1].cpu().numpy()
        val_probs.extend(p)
        val_labels.extend(y.numpy())

val_probs = np.array(val_probs)
val_labels = np.array(val_labels)

best_t, best_f1_t = 0.5, 0
for t in np.arange(0.3,0.8,0.01):
    f1 = f1_score(val_labels,(val_probs>t).astype(int))
    if f1 > best_f1_t:
        best_f1_t, best_t = f1, t

print(f"\nBest Threshold: {best_t:.2f}")

# ================= TEST =================
print("\n🔍 FINAL TEST")

probs, labels = [], []

with torch.no_grad():
    for x,y in test_loader:
        x = x.to(device)
        out = model(x)
        p = torch.softmax(out,1)[:,1].cpu().numpy()
        probs.extend(p)
        labels.extend(y.numpy())

probs = np.array(probs)
labels = np.array(labels)
preds = (probs > best_t).astype(int)

print(classification_report(labels, preds))
print("ROC-AUC:", roc_auc_score(labels, probs))
print("PR-AUC :", average_precision_score(labels, probs))

# ================= SAVE FINAL MODEL =================
torch.save({
    "model_state": model.state_dict(),
    "threshold": best_t,
    "best_val_f1": best_f1,
    "classes": test_ds.classes
}, "final_model.pth")

print("\n✅ Final model saved (final_model.pth)")

# ================= DASHBOARD =================
fig,axes = plt.subplots(4,2,figsize=(14,18))

fpr,tpr,_=roc_curve(labels,probs)
axes[0,0].plot(fpr,tpr); axes[0,0].plot([0,1],[0,1],'--')
axes[0,0].set_title("ROC (Test)")

precision,recall,_=precision_recall_curve(labels,probs)
axes[0,1].plot(recall,precision)
axes[0,1].set_title("PR (Test)")

ths=np.arange(0,1,0.01)
axes[1,0].plot(ths,[f1_score(labels,(probs>t).astype(int)) for t in ths])
axes[1,0].axvline(best_t)
axes[1,0].set_title("F1 vs T")

sns.heatmap(confusion_matrix(labels,preds),annot=True,fmt="d",ax=axes[1,1])
axes[1,1].set_title("Confusion")

axes[2,0].plot(val_losses); axes[2,0].set_title("Val Loss")
axes[2,1].plot(val_f1s); axes[2,1].set_title("Val F1")
axes[3,0].plot(roc_aucs); axes[3,0].set_title("Val ROC-AUC")
axes[3,1].plot(pr_aucs); axes[3,1].set_title("Val PR-AUC")

plt.tight_layout()
plt.show()
```
